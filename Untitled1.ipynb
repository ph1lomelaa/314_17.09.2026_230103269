4. Strided Halo
Packing
Extract/inject non-contiguous left/right columns
and top/bottom rows on the GPU using
dedicated @cuda.jit pack/unpack kernels
into 1D buffers.
Host-side Python loops over
boundary elements or unpinned
synchronous copies are
disallowed.

```python
from numba import cuda
import numpy as np

@cuda.jit
def pack_left_right_kernel(grid, left_buf, right_buf, width, height, halo):
    idx = cuda.grid(1)
    if idx < height:
        left_buf[idx] = grid[idx + halo, halo]
        right_buf[idx] = grid[idx + halo, width + halo - 1]

@cuda.jit
def unpack_left_right_kernel(grid, left_buf, right_buf, width, height, halo):
    idx = cuda.grid(1)
    if idx < height:
        grid[idx + halo, halo - 1] = left_buf[idx]
        grid[idx + halo, width + halo] = right_buf[idx]

@cuda.jit
def pack_top_bottom_kernel(grid, top_buf, bottom_buf, width, height, halo):
    idx = cuda.grid(1)
    if idx < width:
        top_buf[idx] = grid[halo, idx + halo]
        bottom_buf[idx] = grid[height + halo - 1, idx + halo]

@cuda.jit
def unpack_top_bottom_kernel(grid, top_buf, bottom_buf, width, height, halo):
    idx = cuda.grid(1)
    if idx < width:
        grid[halo - 1, idx + halo] = top_buf[idx]
        grid[height + halo, idx + halo] = bottom_buf[idx]
```
